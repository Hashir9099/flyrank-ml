## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Research Question
Which content pages should be reviewed first for refresh based on observable search and engagement signals?

### Unit of Analysis
One content page.

### Decision
Decide which content pages should be prioritized for review and refresh.

### Output
A ranked list of content pages with a refresh opportunity score and supporting reasons.

### Action
Review, update, expand, or improve the highest-ranked pages first.

### Cost of a Wrong Recommendation
If the wrong pages are prioritized, the team may spend time improving pages with little impact while missing pages that could significantly improve search performance and user engagement.

### Why ML Can Help
Machine learning can combine multiple search and engagement signals to rank pages consistently and provide evidence-based recommendations. The final decision remains with human reviewers.
